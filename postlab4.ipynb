```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
housing = fetch_california_housing()
```

```python
housing_df = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)
```

```python
housing_df["MedHouseVal"] = housing.target
```

```python
housing_df.head()
```

```python
housing_df.shape
```

```python
housing_df.info()
```

```python
housing_df.describe()
```

```python
features = [col for col in housing_df.columns if col != "MedHouseVal"]
```

```python
X = housing_df[features]
```

```python
y = housing_df["MedHouseVal"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

# Linear Regression

Linear Regression is a supervised machine learning algorithm used for predicting continuous numerical values. It establishes a linear relationship between independent variables and the target variable by fitting the best possible straight line to the data.

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(X_train, y_train)
```

```python
lr_predictions = lr_model.predict(X_test)
```

```python
lr_mae = mean_absolute_error(y_test, lr_predictions)
```

```python
lr_mse = mean_squared_error(y_test, lr_predictions)
```

```python
lr_rmse = np.sqrt(lr_mse)
```

```python
lr_r2 = r2_score(y_test, lr_predictions)
```

```python
print("Mean Absolute Error :", lr_mae)
print("Mean Squared Error :", lr_mse)
print("Root Mean Squared Error :", lr_rmse)
print("R2 Score :", lr_r2)
```

# Support Vector Regression (SVR)

Support Vector Regression (SVR) is a supervised machine learning algorithm used for regression tasks. It predicts continuous values by finding a hyperplane that minimizes prediction errors while maintaining a specified margin around the data points.

```python
svr_model = SVR(kernel="rbf")
```

```python
svr_model.fit(X_train_scaled, y_train)
```

```python
svr_predictions = svr_model.predict(X_test_scaled)
```

```python
svr_mae = mean_absolute_error(y_test, svr_predictions)
```

```python
svr_mse = mean_squared_error(y_test, svr_predictions)
```

```python
svr_rmse = np.sqrt(svr_mse)
```

```python
svr_r2 = r2_score(y_test, svr_predictions)
```

```python
print("Mean Absolute Error :", svr_mae)
print("Mean Squared Error :", svr_mse)
print("Root Mean Squared Error :", svr_rmse)
print("R2 Score :", svr_r2)
```

# Decision Tree Regression

Decision Tree Regression is a supervised machine learning algorithm that predicts continuous values by recursively splitting the dataset into smaller subsets based on feature values. It forms a tree-like structure to make predictions.

```python
dt_model = DecisionTreeRegressor(random_state=42)
```

```python
dt_model.fit(X_train, y_train)
```

```python
dt_predictions = dt_model.predict(X_test)
```

```python
dt_mae = mean_absolute_error(y_test, dt_predictions)
```

```python
dt_mse = mean_squared_error(y_test, dt_predictions)
```

```python
dt_rmse = np.sqrt(dt_mse)
```

```python
dt_r2 = r2_score(y_test, dt_predictions)
```

```python
print("Mean Absolute Error :", dt_mae)
print("Mean Squared Error :", dt_mse)
print("Root Mean Squared Error :", dt_rmse)
print("R2 Score :", dt_r2)
```
